```python
import pickle

with open('linear_regression_model.pkl', 'rb') as file:
    lr_model = pickle.load(file)

with open('decision_tree_model.pkl', 'rb') as file:
    dt_model = pickle.load(file)

with open('random_forest_model.pkl', 'rb') as file:
    rf_model = pickle.load(file)

print("Models loaded successfully!")
```

```python
import ipywidgets as widgets
from IPython.display import display
import pandas as pd

temperature = widgets.FloatSlider(value=25.0, min=-10.0, max=50.0, step=0.1, description='Temp (°C):')
humidity = widgets.FloatSlider(value=50.0, min=0.0, max=100.0, step=0.1, description='Humidity (%):')
pressure = widgets.FloatSlider(value=1013.25, min=900.0, max=1100.0, step=0.1, description='Pressure (hPa):')
wind_speed = widgets.FloatSlider(value=5.0, min=0.0, max=50.0, step=0.1, description='Wind Speed (km/h):')

output_label = widgets.Label(value="Projected Power Output: N/A")

def update_output(change):
    input_data = pd.DataFrame([[
        temperature.value,
        humidity.value,
        pressure.value,
        wind_speed.value
    ]], columns=["AT", "V", "AP", "RH"])

    prediction = rf_model.predict(input_data)

    output_label.value = f"Projected Power Output: {prediction[0]:.2f} MW"

temperature.observe(update_output, names='value')
humidity.observe(update_output, names='value')
pressure.observe(update_output, names='value')
wind_speed.observe(update_output, names='value')

display(temperature, humidity, pressure, wind_speed)
display(output_label)
```

```python
import ipywidgets as widgets
from IPython.display import display

temperature_input = widgets.FloatText(value=25.0, description='Temp (°C):')
humidity_input = widgets.FloatText(value=50.0, description='Humidity (%):')
pressure_input = widgets.FloatText(value=1013.25, description='Pressure (hPa):')
wind_speed_input = widgets.FloatText(value=5.0, description='Wind Speed (km/h):')

submit_button = widgets.Button(description="Submit")
output_label = widgets.Label(value="Projected Power Output: N/A")

def on_submit_button_click(b):
    input_data = pd.DataFrame([[
        temperature_input.value,
        humidity_input.value,
        pressure_input.value,
        wind_speed_input.value
    ]], columns=["AT", "V", "AP", "RH"])

    prediction = rf_model.predict(input_data)

    output_label.value = f"Projected Power Output: {prediction[0]:.2f} MW"

submit_button.on_click(on_submit_button_click)

display(temperature_input, humidity_input, pressure_input, wind_speed_input, submit_button, output_label)
```
